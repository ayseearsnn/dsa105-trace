```python
vize = float(input("Vize notun: "))
final = float(input("Final notun: "))

ortalama = vize * 0.4 + final * 0.6
print("Ortalaman:", ortalama)

if ortalama >= 50:
    print("Geçtin")
elif ortalama >= 45 and ortalama < 50:
    print("Bütünlemeye kaldı")
else:
    print("Kaldın")
```

```python
revenue = float(input("Revenue"))
expenses = float(input("Expenses"))

profit = revenue - expenses
print("Profit:", profit)

if profit > 0:
    print("Profit")
elif profit < 0:
    print("Loss")
else:
    print("Break even")
```
