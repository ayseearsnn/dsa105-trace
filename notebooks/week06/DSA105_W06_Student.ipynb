## Part A — Core understanding

### 1. Comparisons produce Boolean values


```python
order_total = 1250
free_delivery_threshold = 2000

is_free_delivery = order_total >= free_delivery_threshold

print(is_free_delivery)
print(type(is_free_delivery))
```

The name `is_free_delivery` reads as a yes/no proposition. This is a normal way to name Boolean state.


### 2. Basic branch


```python
order_total = 1250
free_delivery_threshold = 2000

if order_total >= free_delivery_threshold:
    delivery_message = "Free delivery"
else:
    delivery_message = "Standard delivery"

print(delivery_message)
print(type(delivery_message))
```

### 3. Multi-branch rule


```python
order_total = 1500

if order_total >= 2000:
    delivery_message = "Free delivery"
elif order_total >= 1000:
    delivery_message = "Reduced delivery"
else:
    delivery_message = "Standard delivery"

print(delivery_message)
print(type(delivery_message))
```
