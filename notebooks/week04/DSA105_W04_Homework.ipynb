# DSA105 TRACE Homework — Week 4

A service sells:

- Standard plan: 85 subscriptions × 350 TL
- Premium plan: 40 subscriptions × 390 TL

Calculate:

- each plan's revenue;
- total revenue;
- premium revenue share;
- average revenue per subscription.

Write readable expressions with meaningful result names.


```python
standard_subscriptions = 85
standard_price = 350
premium_subscriptions = 40
premium_price = 390

# Complete the calculations.

standard_revenue = standard_subscriptions * standard_price
premium_revenue = premium_subscriptions * premium_price
total_revenue = standard_revenue + premium_revenue
premium_revenue_share = premium_revenue / total_revenue
average_revenue_per_subscription = total_revenue / (standard_subscriptions + premium_subscriptions)

print(standard_revenue)
print(premium_revenue)
print(total_revenue)
print(premium_revenue_share)
print(average_revenue_per_subscription)
```
