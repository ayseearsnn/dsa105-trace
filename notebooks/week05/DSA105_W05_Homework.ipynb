# DSA105 TRACE Homework — Week 5

Build an interactive delivery calculator.

Ask for:

- distance in kilometers;
- price per kilometer.

Calculate the delivery cost and display it with two decimal places.


```python
# Input and conversion

distance_km = float(input("Distance in km: "))
price_per_km = float(input("Price per km: "))
# Computation

distance_km = 300
price_per_km = 100
delivery_cost = distance_km * price_per_km
# Output

print(f"Delivery cost: {delivery_cost:.2f}")
# Output
```
