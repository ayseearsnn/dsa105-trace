```python
import ipywidgets as widgets
from IPython.display import display, clear_output
import pandas as pd

# Global meta inputs
title_input = widgets.Text(value=PLAN_TITLE, description='Plan Title:', layout=widgets.Layout(width='60%'))
prep_input = widgets.Text(value=PREPARATION_PERIOD, description='Prep Period:', layout=widgets.Layout(width='60%'))
deadline_input = widgets.Text(value=EXAM_DEADLINE, description='Exam Range:', layout=widgets.Layout(width='60%'))

output_area = widgets.Output()

# Generate input widgets dynamically for each lesson in study_data
lesson_widgets = []
for idx, row in enumerate(study_data):
    lesson_widgets.append({
        'index': idx,
        'Lesson': widgets.Text(value=row['Lesson'], description=f"L{idx+1} Name:", layout=widgets.Layout(width='25%')),
        'ECTS': widgets.FloatText(value=row['ECTS'], description="ECTS:", layout=widgets.Layout(width='12%')),
        'Study_Dates': widgets.Text(value=row['Study_Dates'], description="Dates:", layout=widgets.Layout(width='20%')),
        'Total_Days': widgets.IntText(value=row['Total_Days'], description="Days:", layout=widgets.Layout(width='10%')),
        'Daily_Duration': widgets.Text(value=row['Daily_Duration'], description="Duration:", layout=widgets.Layout(width='15%')),
        'Study_Place': widgets.Text(value=row['Study_Place'], description="Place:", layout=widgets.Layout(width='15%'))
    })

def update_and_display(change=None):
    global PLAN_TITLE, PREPARATION_PERIOD, EXAM_DEADLINE, study_data
    PLAN_TITLE = title_input.value
    PREPARATION_PERIOD = prep_input.value
    EXAM_DEADLINE = deadline_input.value

    # Update global study_data from widgets
    for item in lesson_widgets:
        idx = item['index']
        study_data[idx]['Lesson'] = item['Lesson'].value
        study_data[idx]['ECTS'] = item['ECTS'].value
        study_data[idx]['Study_Dates'] = item['Study_Dates'].value
        study_data[idx]['Total_Days'] = item['Total_Days'].value
        study_data[idx]['Daily_Duration'] = item['Daily_Duration'].value
        study_data[idx]['Study_Place'] = item['Study_Place'].value

    with output_area:
        clear_output(wait=True)
        display_study_plan(study_data)

# Observe changes
title_input.observe(update_and_display, names='value')
prep_input.observe(update_and_display, names='value')
deadline_input.observe(update_and_display, names='value')

lesson_rows_layout = []
for item in lesson_widgets:
    row_hbox = widgets.HBox([
        item['Lesson'], item['ECTS'], item['Study_Dates'],
        item['Total_Days'], item['Daily_Duration'], item['Study_Place']
    ])
    for widget in [item['Lesson'], item['ECTS'], item['Study_Dates'], item['Total_Days'], item['Daily_Duration'], item['Study_Place']]:
        widget.observe(update_and_display, names='value')
    lesson_rows_layout.append(row_hbox)

print("=== MAIN PARAMETERS ===")
display(title_input, prep_input, deadline_input)

print("\n=== EDIT LESSONS DETAILS ===")
for row in lesson_rows_layout:
    display(row)

display(output_area)

# Initial render
update_and_display()
```
